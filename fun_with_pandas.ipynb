## Music sales challenge

Write a script that:

1. Combines the tables of best-selling physical singles and best-selling digital singles on the Wikipedia page "List_of_best-selling_singles"
2. Adds a column which marks whether each row is from the list of physical singles or digital singles
3. Outputs the artist and single name for the year you were born. If there is no entry for that year, take the closest year after you were born.
4. Outputs the artist and single name for the year you were 15 years old.

```python
# Starter code...

#%pip install lxml
import pandas as pd

rawdata = pd.read_html("https://en.wikipedia.org/wiki/List_of_best-selling_singles")

all_tables_preview = pd.concat(rawdata, keys=range(len(rawdata)))
print(all_tables_preview.head(10))

print(f"Number of tables found: {len(rawdata)}")
for i, table in enumerate(rawdata):
    print(f"\nTable {i}:\n")
    print(table.head())
```

```python
physical_singles = pd.concat([rawdata[0],rawdata[1]])
digital_singles = pd.concat([rawdata[3], rawdata[4], rawdata[5]])

physical_singles['Type'] = 'Physical'
digital_singles['Type'] = 'Digital'

all_singles = pd.concat([physical_singles, digital_singles])

print(all_singles.head(100))

```

```python

birth_year=2000
age_15_year=2015

birth_single = all_singles[all_singles['Released'] == birth_year]
age15_single = all_singles[all_singles['Released'] == age_15_year]

print(birth_single)
print(age15_single)
```

## Space challenge

1. Make a single dataframe that combines the space missions from the 1950's to the 2020's
2. Write a script that returns the year with the most launches
3. Write a script that returns the most common month for launches
4. Write a script that ranks the months from most launches to fewest launches


```python
import pandas as pd

rawdata = pd.read_html("https://en.wikipedia.org/wiki/Timeline_of_Solar_System_exploration")

all_tables_preview = pd.concat(rawdata, keys=range(len(rawdata)))
print(all_tables_preview.head(10))

print(f"Number of tables found: {len(rawdata)}")
for i, table in enumerate(rawdata):
    print(f"\nTable {i}:\n")
    print(table.head())
```

```python
space_data = pd.concat([rawdata[0], rawdata[1], rawdata[2], rawdata[3], rawdata[4], rawdata[5], rawdata[6], rawdata[7],rawdata[8]], ignore_index=True)
space_data['Launch date'] = pd.to_datetime(space_data['Launch date'], errors='coerce')
space_data['Month'] = space_data['Launch date'].dt.month
space_data['Year'] = space_data['Launch date'].dt.year
most_launches_year = space_data['Year'].value_counts().idxmax()
most_common_month = space_data['Month'].value_counts().idxmax()
month_ranking = space_data['Month'].value_counts().sort_values(ascending=False)

print("Year with most launches:", most_launches_year)
print("Most common month:", most_common_month)
print("Month rankings:\n", month_ranking)
```

## Supervillain challenge

1. Write a script that combines the tables showing supervillain debuts from the 30's through the 2010's
2. Write a script that ranks each decade in terms of how many supervillains debuted in that decade
3. Write a script that ranks the different comics companies in terms of how many supervillains they have, and display the results in a nice table (pandas dataframe)

```python
rawdata = pd.read_html("https://en.wikipedia.org/wiki/List_of_comic_book_supervillain_debuts")

all_tables_preview = pd.concat(rawdata, keys=range(len(rawdata)))
print(all_tables_preview.head(10))

print(f"Number of tables found: {len(rawdata)}")
for i, table in enumerate(rawdata):
    print(f"\nTable {i}:\n")
    print(table.head())
```

```python
supervillain_data = pd.concat([rawdata[3], rawdata[4], rawdata[5], rawdata[6], rawdata[7], rawdata[8], rawdata[9], rawdata[10],rawdata[11]], ignore_index=True)

supervillain_data['Year Debuted'] = supervillain_data['Year Debuted'].str.extract(r'(\d{4})').astype(int)
supervillain_data['Decade'] = (supervillain_data['Year Debuted'] // 10) * 10
decade_ranking = supervillain_data['Decade'].value_counts()

company_ranking = supervillain_data['Company'].value_counts()

print("Decade rankings:\n", decade_ranking)
print("Company rankings:\n", company_ranking)
```
